## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

### 2.3 Question Answering

```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
context = (
    "Hugging Face is an AI company with the mission to democratize good machine learning. "
    "The Transformers library provides thousands of pretrained models in 100+ languages."
)

answer = qa({
    "question": "What is the mission of Hugging Face?",
    "context": context
})
print(answer["answer"])
```

### 2.4 Model Discovery with `huggingface_hub`

```python
from huggingface_hub import HfApi

api = HfApi()
# List top 5 English summarization models
models = api.list_models(task="summarization", limit=5)
for m in models:
    print(m.modelId)
```

````python
#sentiment-analysis
from transformers import pipeline
sentiment = pipeline( "sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english" )
````

````python
reviews = [
    "This phone has an excellent camera and the battery lasts all day.",
    "The headphones stopped working after only two days. Very disappointing.",
    "The laptop is good, but the keyboard could be better."]
````

````python
results = sentiment(reviews)
for review, result in zip(reviews, results):
    print(f"Review: {review}")
    print(f"Prediction: {result['label']}")
    print(f"Confidence: {result['score']:.3f}")
    print("-" * 60)
````

````python
test_reviews = [
    "I absolutely love this product. It is worth every penny!",
    "This is the worst product I have ever purchased.",
    "It works as expected and does the job."]
````

````python
results = sentiment(test_reviews)

for review, result in zip(test_reviews, results):
    print(f"{review}")
    print(f"Label: {result['label']}")
    print(f"Score: {result['score']:.3f}")
    print()
````

````python
sentiment_all = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    top_k=None
)

results = sentiment_all([
    "The product is amazing!",
    "The product is terrible."
])

for review, scores in zip(
    ["The product is amazing!", "The product is terrible."],
    results
):
    print(review)
    for score in scores:
        print(f"  {score['label']}: {score['score']:.3f}")
    print()
````

````python
#Text Generation
from transformers import pipeline
generator = pipeline("text-generation", model="gpt2")
````

````python
article = """
Artificial intelligence is becoming an important part of modern life.
Many companies are using AI in healthcare, education, and software
development. AI can help people complete tasks faster and analyze
large amounts of information. However, experts are also concerned
about privacy, misinformation, and job displacement.
"""

summary = generator(
    article,
    max_length=150,
    num_return_sequences=2,
    do_sample=True,
    top_k=50,
    top_p=0.95
)

print("Answer 1:", summary[0]["generated_text"])
print("Answer 2:", summary[1]["generated_text"])
````
